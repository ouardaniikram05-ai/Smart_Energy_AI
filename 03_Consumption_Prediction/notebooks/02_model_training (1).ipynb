```python
# Imports nécessaires au machine learning

import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.linear_model import LinearRegression

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("Libraries imported successfully.")
```

```python
# Chargement des datasets sauvegardés

train_df = pd.read_excel(r"C:\Users\ouard\Smart_Energy_AI\03_Consumption_Prediction\results\train.xlsx")
validation_df = pd.read_excel(r"C:\Users\ouard\Smart_Energy_AI\03_Consumption_Prediction\results\validation.xlsx")
test_df = pd.read_excel(r"C:\Users\ouard\Smart_Energy_AI\03_Consumption_Prediction\results\test.xlsx")

# Conversion temporelle
for dataset in [train_df, validation_df, test_df]:
    dataset["Datetime"] = pd.to_datetime(dataset["Datetime"])

# Variables explicatives et cible
features = [
    "Global_reactive_power",
    "Voltage",
    "Global_intensity",
    "Sub_metering_1",
    "Sub_metering_2",
    "Sub_metering_3"
]

target = "Global_active_power"

X_train = train_df[features]
y_train = train_df[target]

X_val = validation_df[features]
y_val = validation_df[target]

X_test = test_df[features]
y_test = test_df[target]

print("Data loaded and prepared.")
print("Training :", X_train.shape)
print("Validation :", X_val.shape)
print("Testing :", X_test.shape)
```

```python
# Linear Regression

lr_model = LinearRegression()

lr_model.fit(X_train, y_train)

print("Linear Regression training completed.")
```

```python
# Evaluation de Linear Regression sur Validation

y_val_pred_lr = lr_model.predict(X_val)

mae_lr = mean_absolute_error(y_val, y_val_pred_lr)
rmse_lr = np.sqrt(mean_squared_error(y_val, y_val_pred_lr))
r2_lr = r2_score(y_val, y_val_pred_lr)

print("Linear Regression - Validation")
print("MAE :", mae_lr)
print("RMSE :", rmse_lr)
print("R² :", r2_lr)
```

```python
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

print("Random Forest training completed.")
```

```python
# Evaluation de Random Forest sur Validation

y_val_pred_rf = rf_model.predict(X_val)

mae_rf = mean_absolute_error(y_val, y_val_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_val, y_val_pred_rf))
r2_rf = r2_score(y_val, y_val_pred_rf)

print("Random Forest - Validation")
print("MAE :", mae_rf)
print("RMSE :", rmse_rf)
print("R² :", r2_rf)
```

```python
gb_model = GradientBoostingRegressor(
    n_estimators=100,
    random_state=42
)

gb_model.fit(X_train, y_train)

print("Gradient Boosting training completed.")
```

```python
# Evaluation de Gradient Boosting sur Validation

y_val_pred_gb = gb_model.predict(X_val)

mae_gb = mean_absolute_error(y_val, y_val_pred_gb)
rmse_gb = np.sqrt(mean_squared_error(y_val, y_val_pred_gb))
r2_gb = r2_score(y_val, y_val_pred_gb)

print("Gradient Boosting - Validation")
print("MAE :", mae_gb)
print("RMSE :", rmse_gb)
print("R² :", r2_gb)
```

```python
results = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest", "Gradient Boosting"],
    "MAE": [mae_lr, mae_rf, mae_gb],
    "RMSE": [rmse_lr, rmse_rf, rmse_gb],
    "R2": [r2_lr, r2_rf, r2_gb]
})

results = results.sort_values(by="MAE")

results
```

```python
results.to_csv("model_comparison.csv", index=False)

print("Comparison table exported successfully.")
```
